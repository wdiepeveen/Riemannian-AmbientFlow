```python
import torch
import matplotlib.pyplot as plt

data_path = "./data/MNIST/reconstructed/mnist_inpainting_h28_w28_mask0.7_noise0.1_pgd_stepsize[0.99]_g[0.4]_sigma[0.99]_maxiter10"
```

```python
# Plot original, noisy, and denoised images
def show_images(img_list, titles):
    fig, axes = plt.subplots(1, len(img_list), figsize=(12, 4))
    for i, (img, title) in enumerate(zip(img_list, titles)):
        axes[i].imshow(img[0].cpu().numpy().squeeze(), cmap="gray")
        axes[i].set_title(title)
        axes[i].axis("off")
    plt.tight_layout()
    plt.show()
```

```python
save_path = data_path + "/mnist_inpainting_pnp_train.pt"

checkpoint = torch.load(save_path)
cor_data_train = checkpoint['cor_data']
rec_data_train = checkpoint['rec_data']
clean_data_train = checkpoint['clean_data']
start_idx = checkpoint['last_index'] + 1

print(start_idx)
for i in range(5):
    show_images(
        [cor_data_train[i], rec_data_train[i], clean_data_train[i]],
        ["Corrupted", "Reconstructed", "Original"]
    )
```
